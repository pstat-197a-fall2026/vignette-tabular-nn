### Introduction
Tabular neural network (Tabular NN) is a type of neural network designed to learn patterns from structured rows and columns, such as spreadsheets, to do classification and regression tasks. Similar to traditional neural networks, tabular nn works by passing input features through multiple layers of interconnected neurons, which captures complex relationships between variables by adjusting their weights during training.

In this project, we will use a tabular neural network to predict the impact of artificial intelligence on the job market based on the `AI impact level` variable. Through this example, we will walk through the process of data preprocessing, building and training a neural network, tuning its parameters, and evaluating its performance.
